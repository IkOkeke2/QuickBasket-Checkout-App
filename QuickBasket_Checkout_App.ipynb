```python
def collect_items_interactively(cart):
    """
    Ask the cashier to input items info one at a time
    Type 'done' as an item name to stop adding items
    """

    print("=== QUICK BASKET: Enter items below (type 'done' to finish) ===")

    while True:
        # input: collect input for user
        item_name = input("Enter an item: ")
    
        if item_name.lower() =="done":
            break

        try:
            price = float(input("Enter a price: "))
            quantity = int(input("Enter a quantity: "))
        except ValueError:
            print("Please enter numeric values for price and quantity. Try again. \n")
            continue

        if price < 0 or quantity <= 0:
            print("Price must be positive and quantity must be at least 1.\n")
            continue

        add_item(cart, item_name, price, quantity)

    return cart
```

```python
def calculate_subtotal(cart):
    """ Add up price * quantity for every item currently in the cart """
    return sum(item["price"] * item['quantity'] for item in cart.values()) 

def apply_discount(amount, discount_percent=0):
    """ Reduce amount by discount_percent"""
    discount_amount_removed = round(amount * (discount_percent / 100), 2)
    amount_after_discount = amount - discount_amount_removed
    return amount_after_discount, discount_amount_removed

def apply_tax(amount, tax_percent=7.5):
    """ Add tax on top of amount"""
    tax_amount = round(amount * (tax_percent / 100), 2)
    amount_after_tax = amount + tax_amount
    return amount_after_tax, tax_amount

def calculate_totals(cart, discount_percent=0, tax_percent=7.5):
    """ Run the full pricing pipeline: """

    subtotal = calculate_subtotal(cart)
    amount_after_discount, discount_amount = apply_discount(subtotal, discount_percent)
    final_amount, tax_amount = apply_tax(amount_after_discount, tax_percent)

    return{
        "subtotal": subtotal,
        "discount_percent": discount_percent,
        "discount_amount": discount_amount,
        "tax_percent": tax_percent,
        "tax_amount": tax_amount,
        "final_amount": final_amount
    }
```

```python
calculate_totals(cart)
```

```python
calculate_totals(cart, discount_percent=3, tax_percent=4.5)
```

```python
from datetime import datetime 
def generate_receipt(cart, total, store_name="QuickBasket Store", cashier="John"):
    """ Build a formatted, human-readable receipt"""

    width = 42
    row = []

    # store information
    row.append(store_name.center(width))
    row.append("Berlin, Germany".center(width))
    row.append(f"Date: {datetime.now().strftime('%d-%m-%Y %H:%M')}")
    row.append(f"Cashier Name: {cashier}")

    row.append('=' * width)
    row.append(f"{'Item':<18}{'Qty':>4}{'Price($)':>10}{'Total($)':>10}")
    row.append('=' * width)

    for name, details in cart.items():
        item_total = details['price'] * details['quantity']
        row.append(
            f"{name:<18}{details['quantity']:>4}{details['price']:>10,.2f}{item_total:>10,.2f}"
        )

    row.append("=" * width)
    row.append(f"{'Subtotal:':<30}{total['subtotal']:>12,.2f}")

    if total['discount_percent'] > 0:
        label = f"Discount ({total['discount_percent']:.0f}%):"
        row.append(f"{label:<30}{-total['discount_amount']:>12,.2f}")

    tax_label = f"VAT ({total['tax_percent']:.1f}%):"
    row.append(f"{tax_label:<30}{total['tax_amount']:>12,.2f}")

    row.append("=" * width)
    row.append(f"{'TOTAL AMOUNT ($):':<30}{total['final_amount']:>12,.2f}")
    row.append("=" * width)
    row.append("Thank you for shopping with us!".center(width))
    row.append("Come back soon!".center(width))
    
    return row
    
```
